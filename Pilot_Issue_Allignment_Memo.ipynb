# Aarhus Candidate Test & News Articles Idea

## Related Literature: Substantive

1. **Belief Systems & Issue Constraint**
   1. **Converse (1964)**: Voters rely on structured “policy bundles”; issues are not independent but constrained within belief systems.
   1. Implication: Political competition is organized by structured issue bundles, not independent issue stances.
2. **Elite Influence & Party Cues**
   1. **Zaller (1992)**: Elite framing shapes how citizens sample considerations from memory.
   1. Studies on **Party Cues**: Voters use partisan signals to organize issue positions.
   1. Implication: Issue alignment is not only ideological but strategically structured.
3. **Framing & Political Communication**
   1. **Entman (1993)**; **Chong & Druckman (2007)**: Framing defines problem interpretation and evaluative emphasis.
   1. **Lene Aarøe (AU, PoliSci)** (Framing: Audience Frames): How audiences process framed content.
   1. Implication: Framing should be understood beyond single-issue emphasis — potentially as structured, affectively aligned bundles.
   1. note: we could potentially ask for feedback from Lene in the future?

## Related Literature: Methods

1. **Candidate Test & Issue-Based Ideological Measurement**
   1. Roll-call voting based ideology measurement:
     1. **Poole & Rosenthal (NOMINATE / DW-NOMINATE)**
     1. Bayesian IRT roll-call models (Clinton, Jackman & Rivers)
   1. We draw on the tradition of inferring political structure from issue-based responses.
   1. However, instead of estimating a single latent dimension, we examine asymmetric issue weights that structure partisan alignment.
2. **News Articles & Text-as-Data Approaches**
   1. **Media Slant literature** (e.g., Gentzkow & Shapiro)
   1. **Gentzkow, Shapiro & Taddy (2019)**: High-dimensional text differences across groups.
   1. **Rheault & Cochrane (2020)**: Word embeddings for ideological placement.
   1. These approaches primarily estimate actor location in latent ideological space.
   1. Our focus shifts from ideological location to structural configuration and weighted issue alignment.

## Selling Points of This Study: What Makes Our Study Novel?

### 1. From Salience to Politically Discriminating Structure
1. Existing research emphasizes issue salience and perceived importance.
2. We focus on **politically discriminating issues** — those that empirically structure party differentiation and Media's rhetorics.
3. Issue selection is anchored in party alignment, not frequency or subjective prominence.

### 2. From Latent Ideological Location to Asymmetric Issue Weighting
1. Existing research estimates where actors are located in ideological space.
2. We can highlight how specific issues carry **asymmetric structural weights** in defining partisan identity.
3. Not all issues contribute equally to party alignment.

### 3. From Single-Issue Framing to Bundled Structural Framing
1. Framing studies often analyze issues individually.
2. We redefine framing as **networked bundling**, examining:
    1. Issue co-occurrence
    1. Affective (positive/negative) alignment

### 4. From Separate Arenas to a Unified Issue Space
1. Party positioning and media framing are typically studied separately.
2. We place both parties and media within the same issue space:
    1. The same 10 politically discriminating issues
    1. The same sentiment structure (human + LLM labeling (additional work needed))
3. This allows direct structural comparison across arenas.

### 5. Institutional & Contextual Advantage of the Case
1. **Proportional Representation (PR)** reduces convergence pressures and enables parties to sustain asymmetric issue weights.
2. **Local elections** involve tangible and directly feelable policy stakes.
3. This context provides a theoretically advantageous setting to observe structured issue alignment.

### 6. Methodological Contribution
1. We follow established traditions of issue-based ideological measurement and text-as-data analysis.
2. However, we shift the analytical focus:
    1. From ideological location to structure
    1. From frequency to weight
    1. From isolated frames to a bundled package
