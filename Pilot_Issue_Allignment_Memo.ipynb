# Aarhus Candidate Test & News Articles Idea

## Related Literature: Substantive

1. **Belief Systems & Issue Constraint**
   - **Converse (1964)**: Voters rely on structured “policy bundles”; issues are not independent but constrained within belief systems.
   - **Baldassarri & Goldberg (2014)**: Issue alignment emerges as structured constellations rather than isolated positions.
   - Implication: Political competition is organized by structured issue bundles, not independent issue stances.

2. **Elite Influence & Party Cues**
   - **Zaller (1992)**: Elite framing shapes how citizens sample considerations from memory.
   - Studies on **Party Cues**: Voters use partisan signals to organize issue positions.
   - Implication: Issue alignment is not only ideological but strategically structured.

3. **Framing & Political Communication**
   - **Entman (1993)**; **Chong & Druckman (2007)**: Framing defines problem interpretation and evaluative emphasis.
   - **Lene Aarøe (AU, PoliSci)** (Framing: Audience Frames): How audiences process framed content.
   - Implication: Framing should be understood beyond single-issue emphasis — potentially as structured, affectively aligned bundles.
   - note: we could potentially ask for feedback from Lene in the future?

## Related Literature: Methods

1. **Candidate Test & Issue-Based Ideological Measurement**
   - Roll-call voting based ideology measurement:
     - **Poole & Rosenthal (NOMINATE / DW-NOMINATE)**
     - Bayesian IRT roll-call models (Clinton, Jackman & Rivers)
   - We draw on the tradition of inferring political structure from issue-based responses.
   - However, instead of estimating a single latent dimension, we examine asymmetric issue weights that structure partisan alignment.

2. **News Articles & Text-as-Data Approaches**
   - **Media Slant literature** (e.g., Gentzkow & Shapiro)
   - **Gentzkow, Shapiro & Taddy (2019)**: High-dimensional text differences across groups.
   - **Rheault & Cochrane (2020)**: Word embeddings for ideological placement.
   - These approaches primarily estimate actor location in latent ideological space.
   - Our focus shifts from ideological location to structural configuration and weighted issue alignment.

## Selling Points of This Study: What Makes Our Study Novel?

### 1. From Salience to Politically Discriminating Structure
- Existing research emphasizes issue salience and perceived importance.
- We focus on **politically discriminating issues** — those that empirically structure party differentiation and Media's rhetorics.
- Issue selection is anchored in party alignment, not frequency or subjective prominence.

### 2. From Latent Ideological Location to Asymmetric Issue Weighting
- Existing research estimates where actors are located in ideological space.
- We can highlight how specific issues carry **asymmetric structural weights** in defining partisan identity.
- Not all issues contribute equally to party alignment.

### 3. From Single-Issue Framing to Bundled Structural Framing
- Framing studies often analyze issues individually.
- We redefine framing as **networked bundling**, examining:
  - Issue co-occurrence
  - Affective (positive/negative) alignment

### 4. From Separate Arenas to a Unified Issue Space
- Party positioning and media framing are typically studied separately.
- We place both parties and media within the same issue space:
  - The same 10 politically discriminating issues
  - The same sentiment structure (human + LLM labeling (additional work needed))
- This allows direct structural comparison across arenas.

### 5. Institutional & Contextual Advantage of the Case
- **Proportional Representation (PR)** reduces convergence pressures and enables parties to sustain asymmetric issue weights.
- **Local elections** involve tangible and directly feelable policy stakes.
- This context provides a theoretically advantageous setting to observe structured issue alignment.

### 6. Methodological Contribution
- We follow established traditions of issue-based ideological measurement and text-as-data analysis.
- However, we shift the analytical focus:
  - From ideological location to structure
  - From frequency to weight
  - From isolated frames to a bundled package
